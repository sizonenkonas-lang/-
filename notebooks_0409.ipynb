Задание 7.
Строим гистограмму

```python
plt.figure(figsize=(12, 6))
plt.hist(df['popularity'], bins=20, edgecolor='black', alpha=0.7, color='steelblue')

plt.title('Распределение значений popularity', fontsize=16, fontweight='bold', pad=15)
plt.xlabel('Popularity', fontsize=12)
plt.ylabel('Количество треков', fontsize=12)

plt.xlim(0, 100)
plt.xticks(range(0, 101, 10))
plt.grid(axis='y', alpha=0.5)

mean_val = df['popularity'].mean()
median_val = df['popularity'].median()

plt.axvline(mean_val, color='red', linestyle='--', linewidth=2, label=f'Среднее: {mean_val:.1f}')
plt.axvline(median_val, color='green', linestyle='--', linewidth=2, label=f'Медиана: {median_val:.1f}')
plt.legend()

plt.tight_layout()
plt.show()
```

Основная часть значений в диапазоне от 14 до 19 и от 23 до 28. Значительно реже в диапазоне от 75 до 100.

Контрольные вопросы: 
1.DataFrame — это двумерная табличная структура данных в библиотеке Pandas, похожая на таблицу в Excel или SQL. Она состоит из строк и столбцов, где каждый столбец может содержать данные разных типов.
2.Признак — это отдельный столбец в наборе данных, который описывает определённую характеристику объекта. Например, в нашем датасете признаками являются: popularity, genre, stream_count, release_year и т.д.
3.Среднее арифметическое — сумма всех значений, делённая на их количество. Чувствительно к выбросам. Медиана — значение, которое находится в центре отсортированного ряда данных. Устойчива к выбросам.
4.Гистограмма показывает распределение числовых данных — как часто встречаются значения в определённых интервалах (бинаx). Позволяет оценить форму распределения, наличие выбросов и концентрацию данных.
5.Фильтрация данных нужна для того, чтобы выбрать из всего набора только те строки, которые соответствуют определённым условиям. Это позволяет анализировать подмножества данных (например, только треки после 2024 года или только определённого жанра).
